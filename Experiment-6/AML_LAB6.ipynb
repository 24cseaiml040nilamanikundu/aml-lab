```python
#implement the following dataset where sports is related to target variable buyes protien powder.consider the level of significance alpha=0.05 which xhie-square tabulated valued is 3.841. apply chie-square of goodness fit in a table.

import numpy as np

observed=np.array([
    [10,5],[5,10]
])

chi_square_tabulated=3.841
print("Chi-square Tabulated Value:",chi_square_tabulated)

print("Observed Value are:")
print(observed)

level_of_significance=0.05
print("Level of Significance:",level_of_significance)

#calculated expected value
expected=np.outer(np.sum(observed,axis=1),np.sum(observed,axis=0))/np.sum(observed)
print("Expected Value are:")
print(expected)

#calculate chi-square
chi_square=np.sum((observed-expected)**2/expected)
print("Chi-square Value:",chi_square)

#calculate degree of freedom
df=(observed.shape[0]-1)*(observed.shape[1]-1)
print("Degree of Freedom:",df)

if chi_square>chi_square_tabulated:
    print("Here x^2 calculated > x^2 tablulatedReject the null hypothesis")
else:
    print("Here x^2 calculated < x^2 tablulated,then Accept the null hypothesis")
```
